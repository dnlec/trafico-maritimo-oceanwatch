# Configuración compartida

Constantes del catálogo y el widget `fecha`. Los demás notebooks lo incluyen
con `%run ./_config`. El job pasa `fecha` y rellena el widget.


```python
import datetime as dt

spark.conf.set("spark.sql.session.timeZone", "UTC")

CATALOG = "dbr_acad_ingsistemas_mine"
SCHEMA = "default"
VOLUME = "oceanwatch_data"

VOLUME_NAME = f"`{CATALOG}`.`{SCHEMA}`.`{VOLUME}`"
VOLUME_ROOT = f"/Volumes/{CATALOG}/{SCHEMA}/{VOLUME}"
RAW_ZIPS = f"{VOLUME_ROOT}/raw_zips"
LANDING = f"{VOLUME_ROOT}/landing"
CHECKPOINTS = f"{VOLUME_ROOT}/checkpoints"
SCHEMAS = f"{VOLUME_ROOT}/_schemas"
REF = f"{VOLUME_ROOT}/ref"
STAGING = f"{RAW_ZIPS}/_staging"
CHECKPOINT_BRONZE = f"{CHECKPOINTS}/bronze_ais"
SCHEMA_LOC = f"{SCHEMAS}/bronze_ais"
CHECKPOINT_SILVER = f"{CHECKPOINTS}/silver_ais_v2"

BRONZE = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_bronze_ais`"
SILVER = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_silver_ais_v2`"
CUARENTENA = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_cuarentena_v2`"
METRICAS = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_metricas`"
PUERTOS = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_puertos`"
GOLD_TRAY = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_gold_trayectorias`"
GOLD_PUERTO = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_gold_presencia_puertos`"
GOLD_TRAFICO = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_gold_trafico_celda_hora`"

GOLD_ANOM = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_gold_anomalias`"
METRICAS_V2 = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_metricas_silver_v2`"
CONTROL_ORO = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_control_oro_v2`"
MODELO = f"{CATALOG}.{SCHEMA}.oceanwatch_detector_velocidad"

fecha, ancla, inicio_run = globals().get("ow_argumentos", ("2023-06-08", "", ""))
if fecha.upper() == "AUTO":
    # Ancla fija del replay y fecha UTC del inicio del run, estables en sus reintentos.
    avance = (dt.date.fromisoformat(inicio_run) - dt.date.fromisoformat(ancla)).days
    if avance < 0:
        raise ValueError("El inicio del run precede al ancla del replay.")
    fecha_dia = dt.date(2023, 6, 8) + dt.timedelta(days=avance)
else:
    fecha_dia = dt.date.fromisoformat(fecha)
fecha = fecha_dia.isoformat()
print(f"OceanWatch · fecha AIS {fecha} UTC · {VOLUME_ROOT}")
CONTROL_ML = f"`{CATALOG}`.`{SCHEMA}`.`oceanwatch_control_ml_v2`"
```
